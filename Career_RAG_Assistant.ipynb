```python
from pypdf import PdfReader
from sentence_transformers import SentenceTransformer
import faiss
import numpy as np

print("Libraries Loaded")
```

```python
def extract_pdf_text(pdf_file):

    reader = PdfReader(pdf_file)

    text = ""

    for page in reader.pages:

        extracted = page.extract_text()

        if extracted:
            text += extracted

    return text
```

```python
resume_file = list(uploaded.keys())[0]

resume_text = extract_pdf_text(resume_file)

jd_text = """
Machine Learning Engineer

Required Skills:
Python
Machine Learning
TensorFlow
Docker
AWS
SQL

Responsibilities:
Develop machine learning models.
Deploy AI applications.
Work with cloud technologies.
"""

print("Resume Loaded")
print("Sample JD Loaded")
```

```python
def create_chunks(text):

    words = text.split()

    chunks = []

    chunk_size = 100

    for i in range(0, len(words), chunk_size):

        chunk = " ".join(words[i:i+chunk_size])

        chunks.append(chunk)

    return chunks
```

```python
resume_chunks = create_chunks(
    resume_text
)

jd_chunks = create_chunks(
    jd_text
)

all_chunks = resume_chunks + jd_chunks

print("Resume Chunks:", len(resume_chunks))
print("JD Chunks:", len(jd_chunks))
print("Total Chunks:", len(all_chunks))
```

```python
model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)

embeddings = model.encode(
    all_chunks
)

print("Embeddings Created")
print(embeddings.shape)
```

```python
dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(
    dimension
)

index.add(
    np.array(embeddings)
)

print("Vector Database Created")
print("Total Vectors:", index.ntotal)
```

```python
def search(query):

    query_embedding = model.encode(
        [query]
    )

    D, I = index.search(
        np.array(query_embedding),
        3
    )

    results = []

    for idx in I[0]:
        results.append(
            all_chunks[idx]
        )

    return "\n".join(results)
```

```python
query = input(
    "Ask Career Assistant: "
)

result = search(query)

print("\nRetrieved Information:\n")
print(result)
```

```python
skills = [
    "python",
    "machine learning",
    "tensorflow",
    "docker",
    "aws",
    "sql"
]

resume_skills = []

for skill in skills:

    if skill in resume_text.lower():
        resume_skills.append(skill)

missing = []

for skill in skills:

    if skill in jd_text.lower() and skill not in resume_skills:
        missing.append(skill)

print("Resume Skills:")
print(resume_skills)

print("\nMissing Skills:")
print(missing)
```

```python
match_score = (
    (len(skills) - len(missing))
    / len(skills)
) * 100

print(
    "Resume Match Score:",
    round(match_score,2),
    "%"
)
```

```python
print("="*50)
print("PERSONALIZED LEARNING ROADMAP")
print("="*50)

month = 1

for skill in missing:

    print(
        f"Month {month}: Learn {skill.title()}"
    )

    month += 1
```

```python
roles = {

    "Data Scientist":[
        "python",
        "sql",
        "machine learning"
    ],

    "Machine Learning Engineer":[
        "python",
        "machine learning",
        "tensorflow",
        "docker"
    ],

    "AI Engineer":[
        "python",
        "tensorflow",
        "aws"
    ],

    "Backend Developer":[
        "python",
        "sql"
    ]

}

scores = {}

for role, role_skills in roles.items():

    matched = len(
        set(resume_skills)
        &
        set(role_skills)
    )

    score = (
        matched /
        len(role_skills)
    ) * 100

    scores[role] = score

best_role = max(
    scores,
    key=scores.get
)

print(
    "Recommended Career:",
    best_role
)

print(
    "Career Match:",
    round(scores[best_role],2),
    "%"
)
```
